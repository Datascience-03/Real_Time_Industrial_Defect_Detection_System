# Demo: Run inference with the trained YOLO model
This notebook demonstrates loading the exported ONNX model and running inference on a sample test image.

```python
from ultralytics import YOLO
from IPython.display import Image, display

# Adjust path if needed
model_path = 'runs/detect/train/weights/best.onnx'
img_path = 'dataset/test/images/rolled-in_scale_277.jpg'

model = YOLO(model_path)
results = model.predict(source=img_path, conf=0.25, imgsz=640)
# Save annotated result
out_name = 'demo_annotated.jpg'
results[0].save(filename=out_name)
display(Image(out_name))
```

Notes:
- Install `ultralytics` and other dependencies in the environment before running.
- To run with ONNX Runtime GPU, install `onnxruntime-gpu`.
